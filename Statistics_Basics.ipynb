**Завдання 1.** Велика літера Сігма грецького алфавіту позначає в математиці суму. Знайдіть з використанням Python $\Sigma x$, $\Sigma x^2$, та $\Sigma (x-1)^2$ для наступного набору даних:
`1 3 4`
Виведіть результат кожного обчислення на екран.

```python
# Медіана без використання жодних бібліотек, користуючись тільки Python
sorted_purchases = sorted(sum_purchases)
n = len(sorted_purchases)

median = (sorted_purchases[n // 2] + sorted_purchases[(n - 1) // 2]) / 2
print("Медіана (без використання жодних бібліотек):", median)
```

```python
# Медіана використовуючи відповідний метод бібліотеки numpy
median = np.median(sum_purchases)
print("Медіана (використання бібліотеки numpy):", median)
```

Значення вибіркового середнього дорівнювали 667.95, тоді як Медіана 319.0. Вибіркове середнє (667.95) суттєво перевищує медіану (319.0). У вибірці наявні суттєві викиди — значення 4310 та 3524, які сильно затягують середнє значення вгору, оскільки середнє розраховується з урахуванням суми всіх елементів. Медіана ж є стійкою до таких викидів.

Вибіркове середнє краще використовувати для *симетричних даних без викидів*, де дані розподілені відносно рівномірно навколо центру (наприклад, зріст людей, температура, вага готової продукції).

Медіану краще використовувати для *асиметричних даних чи даних із викидами* або економічних показників, таких як доходи / зарплати населення.

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.  
  4.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран.

```python
# Стандартне відхилення без використання жодних бібліотек, користуючись тільки Python
n = len(sum_purchases)
mean_val = sum(sum_purchases) / n

# Для вибірки використовуємо формулу незміщеної дисперсії (ділення на n - 1)
variance_sample = sum((x - mean_val) ** 2 for x in sum_purchases) / (n - 1)
std_pure = variance_sample ** 0.5

print("Стандартне відхилення (без використання бібліотек):", std_pure)
```

```python
# Стандартне відхилення використовуючи відповідний метод бібліотеки numpy
# Вказуємо ddof=1 для вибіркового стандартного відхилення
std_np = np.std(sum_purchases, ddof=1)
print("Стандартне відхилення (використання бібліотеки numpy):", std_np)
```

**Завдання 5**. Ми провели дослідження віку студентів-першокурсників. Виявилось, що
- 10 студентів мають вік 16 років
- 25 студентів - 17 років
- 20 студентів - 18 років
- 2 студенти - 20 років

Обчислість середнє значення для цієї вибірки будь-яким способом - на чистому Python або з numpy - та виведіть результат на екран.

```python
# За допомогою Python

# Словник: {вік: кількість студентів}
students_age = {
    16: 10,
    17: 25,
    18: 20,
    20: 2}

# Сумарна кількість студентів (загальний розмір вибірки)
total_students = sum(students_age.values())

# Загальна сума років усіх студентів
total_years = sum(age * count for age, count in students_age.items())

# Середній вік
mean_age = total_years / total_students

print("Загальна кількість студентів:", total_students)
print(f"Середній вік студентів: {mean_age:.2f} років")
```

```python
# За допомогою numpy

# Створюємо масиви для віку та кількості студентів
ages = np.array([16, 17, 18, 20])
counts = np.array([10, 25, 20, 2])

# Обчислюємо середнє за допомогою numpy.average
mean_age_np = np.average(ages, weights=counts)

print(f"Середній вік студентів (numpy): {mean_age_np:.2f}")
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
# Коефіцієнт кореляції Пірсона без використання бібліотек (Python)
n = len(sum_purchases)

mean_x = sum(sum_purchases) / n
mean_y = sum(visits) / n

cov_xy = sum((x - mean_x) * (y - mean_y) for x, y in zip(sum_purchases, visits))
std_x = (sum((x - mean_x) ** 2 for x in sum_purchases)) ** 0.5
std_y = (sum((y - mean_y) ** 2 for y in visits)) ** 0.5

r_pure = cov_xy / (std_x * std_y)

# Коефіцієнт кореляції Пірсона з використанням numpy
r_matrix = np.corrcoef(sum_purchases, visits)
r_np = r_matrix[0, 1]

print(f"Коефіцієнт кореляції Пірсона (Python): {r_pure:.2f}")
print(f"Коефіцієнт кореляції Пірсона (numpy): {r_np:.2f}")
```

Значення кореляції $r \approx -0.30$ вказує на від'ємну (обернену) залежність. Тобто за цими даними спостерігається слабка тенденція: *чим більше відвідувань робить покупець, тим меншою є загальна сума його покупок (і навпаки)*.
Оскільки за модулем $\vert{}r\vert{} \approx 0.30$ знаходиться в межах від $0.30$ до $0.50$, це свідчить про *слабкий (або дуже помірний) лінійний зв'язок між кількістю відвідувань та сумою покупок*.

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

Основна маса даних не утворює чіткої лінійної структури, а нагадує вертикальну «хмару» або стійке скупчення в діапазоні низьких сум покупок.

На графіку добре видно дві точки праворуч із великими значеннями по осі X (понад 3500–4000 одиниць). Це викиди (значення 4310 та 3524), яким відповідають досить невеликі значення відвідувань (4 та 2). Саме ці два покупці сильно зсувають значення кореляції в бік від'ємного зв'язку. Якщо прибрати ці два аномальні спостереження, основна маса даних на графіку покаже, що лінійного зв'язку між кількістю відвідувань і сумою покупок практично немає (точки розподілені хаотично вертикально).

Якщо відкидати дані не доцільно з математичних міркувань, замість коефіцієнта Пірсона можна використати коефіцієнт рангової кореляції Спірмена (scipy.stats.spearmanr), оскільки він набагато стійкіший до викидів і оцінює монотонність зв'язку, а не суто лінійну залежність.

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
# Перетворюємо списки на масиви numpy
sum_purchases = np.array([
    316, 493, 367, 153, 224, 121, 322, 450, 196, 397,
    281, 239, 484, 124, 223, 448, 432, 255, 4310, 3524
])
visits = np.array([
    6, 2, 10, 3, 9, 9, 2, 3, 6, 6,
    10, 2, 5, 10, 3, 5, 4, 7, 4, 2
])

# Обчислюємо порогове значення
mean_sp = np.mean(sum_purchases)
std_sp = np.std(sum_purchases, ddof=1)
threshold = mean_sp + 2 * std_sp

# Фільтруємо масиви за допомогою булевої маски
mask = sum_purchases <= threshold
sum_purchases_clean = sum_purchases[mask]
visits_clean = visits[mask]

# Обчислюємо коефіцієнт кореляції Пірсона
r_clean = np.corrcoef(sum_purchases_clean, visits_clean)[0, 1]

print(f"Коефіцієнт кореляції Пірсона (без викидів): {r_clean:.4f}")

```

```python
# Побудова діаграми розсіювання

plt.figure(figsize=(7, 5))
plt.scatter(sum_purchases_clean, visits_clean, color='blue', alpha=0.7, edgecolors='k')
plt.title('Scatter Plot of Purchases vs. Visits (Without Outliers)')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show();
```
